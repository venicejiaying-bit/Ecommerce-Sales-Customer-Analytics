```python
import pandas as pd
import matplotlib.pyplot as plt
import sqlite3
import os
```

```python
df = pd.read_excel("../data/raw/superstore_sales.csv.xlsx")

df.head()
```

```python
os.makedirs("../data/processed", exist_ok=True)

connection = sqlite3.connect("../data/processed/sales.db")

print("Database connection successful!")
```

```python
df.to_sql(
    "sales",
    connection,
    if_exists="replace",
    index=False
)

print("Sales data imported successfully!")
```

```python
query = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql_query(query, connection)

result
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
print("Overall Profit Margin:", round(profit_margin, 2), "%")
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

df.to_csv(
    "../data/processed/superstore_sales_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
df["Month"] = df["Order Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales
```

```python
monthly_sales.plot(kind="line", figsize=(12, 5), marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
subcategory_analysis = df.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

subcategory_analysis["Profit_Margin_%"] = (
    subcategory_analysis["Total_Profit"]
    / subcategory_analysis["Total_Sales"]
    * 100
)

subcategory_analysis = subcategory_analysis.sort_values(
    by="Total_Profit"
)

subcategory_analysis.round(2)
```

```python
subcategory_analysis["Total_Profit"].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Total Profit by Product Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45, ha="right")
plt.axhline(y=0, linewidth=1)
plt.tight_layout()
plt.show()
```

```python
discount_analysis = df.groupby("Discount").agg(
    Number_of_Orders=("Order ID", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

discount_analysis["Profit_Margin_%"] = (
    discount_analysis["Total_Profit"]
    / discount_analysis["Total_Sales"]
    * 100
)

discount_analysis.round(2)
```

```python
query = """
SELECT *
FROM sales
LIMIT 5;
"""

result = pd.read_sql_query(query, connection)

result
```

```python
query = """
SELECT
    Category,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    SUM(Quantity) AS Total_Quantity
FROM sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

category_sql = pd.read_sql_query(query, connection)

category_sql
```

```python
query = """
SELECT
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

top_products = pd.read_sql_query(query, connection)

top_products
```

```python
query = """
SELECT
    "Customer Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    COUNT(DISTINCT "Order ID") AS Number_of_Orders
FROM sales
GROUP BY "Customer ID", "Customer Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

top_customers = pd.read_sql_query(query, connection)

top_customers
```

```python
query = """
SELECT
    Region,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    SUM(Quantity) AS Total_Quantity
FROM sales
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

regional_analysis = pd.read_sql_query(query, connection)

regional_analysis
```

```python
# Calculate total sales for each category
category_sales = df.groupby("Category")["Sales"].sum()

# Create the bar chart
category_sales.plot(kind="bar", figsize=(8, 5))

# Add chart title and labels
plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)

# Display the chart
plt.tight_layout()
plt.show()
```

```python
# Calculate total profit for each sub-category
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum()

# Sort from lowest to highest profit
subcategory_profit = subcategory_profit.sort_values()

# Create the bar chart
subcategory_profit.plot(kind="bar", figsize=(12, 6))

# Add chart title and labels
plt.title("Total Profit by Sub-Category")
plt.xlabel("Product Sub-Category")
plt.ylabel("Total Profit ($)")

# Add a horizontal line at zero
plt.axhline(y=0, color="black", linewidth=0.8)

# Display the chart
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Group sales by month
monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

# Create the line chart
monthly_sales.plot(
    kind="line",
    figsize=(12, 5),
    marker="o"
)

# Add chart title and labels
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")

# Display the chart
plt.tight_layout()
plt.show()
```

```python
# Calculate total sales for each region
regional_sales = df.groupby("Region")["Sales"].sum()

# Sort from highest to lowest sales
regional_sales = regional_sales.sort_values(ascending=False)

# Create the bar chart
regional_sales.plot(kind="bar", figsize=(8, 5))

# Add chart title and labels
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)

# Display the chart
plt.tight_layout()
plt.show()
```
